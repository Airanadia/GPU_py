## 1. Conceptual Understanding

**Why is a GPU more efficient than a CPU for training deep learning models?**

A GPU is more efficient than a CPU for deep learning because it has many processing cores that can perform multiple calculations simultaneously. This allows it to process large amounts of data faster, reducing the time required to train neural networks.

**What types of operations are best handled by a GPU in this context?**

GPUs are particularly effective at handling parallel mathematical operations, especially matrix multiplication, tensor operations, and other numerical calculations used during neural network training.

```python
import torch

print("CUDA Available:", torch.cuda.is_available())

print("GPU Device:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU detected")
```
